```python
## create a simple text document
import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_texts={
    "../data/text_files/python_intro.txt": """This is the first document. It contains some sample text for testing.
    
    Please note that this is a simple example and can be expanded with more content. Create additional documents as needed for your testing purposes.
    Python is a versatile programming language that can be used for a wide range of applications, including web development, data analysis, artificial intelligence, and more. It is known for its simplicity and readability, making it a popular choice among developers of all skill levels.
    key features of Python include its extensive standard library, support for multiple programming paradigms, and a large community that contributes to its growth and development. Whether you are a beginner or an experienced programmer, Python offers a wealth of resources and tools to help you achieve your goals.
    """,

    "../data/text_files/python_advanced.txt": """This is the second document. It contains more advanced topics in Python programming.
    In this document, we will explore some of the more advanced features of Python, including decorators, generators, and context managers. These features allow developers to write more efficient and maintainable code, and can help to simplify complex programming tasks.
    Decorators are a powerful feature of Python that allow you to modify the behavior of functions or classes. They can be used to add functionality, enforce rules, or perform other tasks before or after the execution of a function or method.
    """
}

for file_path, content in sample_texts.items():
    with open(file_path, "w", encoding="utf-8") as f:
        f.write(content)
print("Sample text documents created successfully!")
```

```python
###Text loader
# from langchain.document_loaders import TextLoader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
loader
document = loader.load()
print(document)
```

```python
### Directory loader
from langchain_community.document_loaders import DirectoryLoader

# Load all text files from the specified directory
dir_loader = DirectoryLoader(
    "../data/text_files", 
    glob="**/*.txt", ## Pattern to match all text files in the directory and its subdirectories
    loader_cls=TextLoader, ## Specify the loader class to use for loading the files
    loader_kwargs={"encoding": "utf-8"},  ## Additional arguments to pass to the loader class
    show_progress=False  ## Show progress while loading the files
)

documents = dir_loader.load()
documents
```
